# Week 1 – Python for Data Science Basics

## Student Marks Dataset Analysis

### Objective

The objective of this project is to analyze the Student Marks Dataset using Python, Pandas, and NumPy. The project focuses on basic data exploration, data cleaning, and statistical analysis of student scores.


## 1. Import Libraries

```python
import pandas as pd
import numpy as np
```

## 2. Load Dataset

The Student Marks Dataset was loaded from a CSV file using Pandas.

```python
df = pd.read_csv("https://raw.githubusercontent.com/SWARGAMVINAY/week1-python-data-science/refs/heads/main/StudentsPerformance.csv")

df.head()
```

## 3. Explore the Dataset

```python
df.shape
```

```python
df.info()
```

## 4. Data Cleaning

Before performing the analysis, the dataset was checked for missing values and duplicate records.

### 4.1 Missing Value Check

The dataset was checked for missing values using `isnull().sum()`.

```python
df.isnull().sum()
```

No missing values were found in any column.

## Score Analysis

We analyze the Math, Reading, and Writing scores to calculate their average, maximum, and minimum values.

```python
average_scores = df[['math score', 'reading score', 'writing score']].mean()

average_scores
```

```python
maximum_scores = df[['math score', 'reading score', 'writing score']].max()

maximum_scores
```

```python
minimum_scores = df[['math score', 'reading score', 'writing score']].min()

minimum_scores
```

## 6. Score Summary

```python
score_summary = pd.DataFrame({
    'Average': average_scores,
    'Minimum': minimum_scores,
    'Maximum': maximum_scores
})

score_summary
```

## 7. Statistical Summary

The `describe()` method provides descriptive statistics for the numerical score columns, including count, mean, standard deviation, minimum, quartiles, and maximum values.

```python
df.describe()
```

## 8. Score Comparison

The following chart compares the average scores across Mathematics, Reading, and Writing.

```python
import matplotlib.pyplot as plt

average_scores.plot(kind='bar')

plt.title('Average Student Scores')
plt.xlabel('Subject')
plt.ylabel('Average Score')
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.show()
```

## 9. Conclusion

The Student Marks Dataset containing 1,000 student records was successfully analyzed using Python, Pandas, and NumPy.

The dataset was inspected for missing values and duplicate records. No missing values or duplicate records were found.

The average scores were 66.089 for Mathematics, 69.169 for Reading, and 68.054 for Writing. The minimum and maximum scores were also calculated for each subject.

This project provided practical experience with loading a CSV dataset, inspecting data, performing basic data cleaning, calculating descriptive statistics, and presenting results using Pandas.
